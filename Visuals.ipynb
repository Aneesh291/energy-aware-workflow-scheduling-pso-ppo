```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("results/four_model_comparison.csv")

workflow_sizes = [25, 50, 100]

plt.figure()
plt.plot(workflow_sizes, df["PSO_Mean"], marker='o', label="PSO Only")
plt.plot(workflow_sizes, df["PPO_Mean"], marker='s', label="PPO Only")
plt.plot(workflow_sizes, df["PPO_DVFS_Mean"], marker='^', label="PPO + DVFS")
plt.plot(workflow_sizes, df["Hybrid_Mean"], marker='D', label="True Hybrid")

plt.xlabel("Number of Tasks")
plt.ylabel("Mean Energy (J)")
plt.title("Four-Model Energy Comparison")
plt.legend()
plt.grid(True)
plt.savefig("results/four_model_energy_vs_size.png")
```

```python
plt.figure()

plt.plot(workflow_sizes,
         df["PSO_Mean"] / workflow_sizes,
         marker='o', label="PSO Only")

plt.plot(workflow_sizes,
         df["PPO_Mean"] / workflow_sizes,
         marker='s', label="PPO Only")

plt.plot(workflow_sizes,
         df["PPO_DVFS_Mean"] / workflow_sizes,
         marker='^', label="PPO + DVFS")

plt.plot(workflow_sizes,
         df["Hybrid_Mean"] / workflow_sizes,
         marker='D', label="True Hybrid")

plt.xlabel("Number of Tasks")
plt.ylabel("Energy per Task (J)")
plt.title("Normalized Energy Consumption per Task")
plt.legend()
plt.grid(True)
plt.savefig("results/normalized_energy_per_task.png")
```

```python
improvement = (df["PSO_Mean"] - df["Hybrid_Mean"]) / df["PSO_Mean"] * 100

plt.figure()
plt.plot(workflow_sizes, improvement, marker='o')

plt.xlabel("Number of Tasks")
plt.ylabel("Improvement over PSO (%)")
plt.title("Hybrid Improvement over PSO Baseline")
plt.grid(True)
plt.savefig("results/hybrid_improvement_percentage.png")
```

```python
x = np.arange(len(workflow_sizes))
width = 0.2

plt.figure(figsize=(10,6))

plt.bar(x - 1.5*width, df["PSO_Mean"], width, label="PSO")
plt.bar(x - 0.5*width, df["PPO_Mean"], width, label="PPO")
plt.bar(x + 0.5*width, df["PPO_DVFS_Mean"], width, label="PPO+DVFS")
plt.bar(x + 1.5*width, df["Hybrid_Mean"], width, label="Hybrid")

plt.xticks(x, workflow_sizes)
plt.xlabel("Workflow Size")
plt.ylabel("Mean Energy (J)")
plt.title("Energy Comparison Across Models")
plt.legend(loc="upper left", frameon=True)
plt.savefig("results/four_model_bar_comparison.png")
```
